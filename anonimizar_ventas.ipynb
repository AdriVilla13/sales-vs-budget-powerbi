```python
from openpyxl import load_workbook
```

```python
FACTOR = #Escoge tu factor  

wb = load_workbook("Ventas.xlsx")
ws = wb["Ventas"]
```

```python
encabezados = [c.value for c in ws[1]]
cols = [i for i, h in enumerate(encabezados, start=1) if str(h).startswith("venta_")]
for fila in ws.iter_rows(min_row=2):
    for i in cols:
        celda = fila[i - 1]
        if isinstance(celda.value, (int, float)):
            celda.value = celda.value * FACTOR
```

```python
wb.save("Ventas_anonimizado.xlsx")
```

```python
ws = wb["Proyectado"]
for fila in ws.iter_rows():
    for celda in fila:
        if isinstance(celda.value, (int, float)) and not (1900 < celda.value < 2100):
            celda.value = celda.value * FACTOR
wb.save("Presupuesto_anonimizado.xlsx")
```
